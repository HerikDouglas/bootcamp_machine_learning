**1. Explique, com suas palavras, o que é machine learning?**

O conceito de machine learning se refere às técnicas empregadas para desenvolver modelos computacionais capazes de identificar padrões dados pré-existentes e aprender com eles, de forma que se tornem capazes de fazer classificações, predições e outras tarefas ao receber novos dados ainda não conhecidos, sem nencessidade de serem explicitamente programados para cada situação.

**2. Explique o conceito de conjunto de treinamento, conjunto de validação e conjunto de teste em machine learning.**

O conjunto de treinamento é uma porcentagem do conjunto de dados utilizado na construção do modelo, geralmente em torno de 70%, para que ele identifique os padrões relevantes a serem empregados em seu aprendizado, é o conjunto de treino oficial do modelo. O conjunto de validação, outra porcentagem dos dados que estão sendo utilizados para a construção do modelo, ajuda a avaliar o desempenho do algoritmo, identificando se ele está tendendo excessivamente para o conjunto de treinamento ou com dificuladades de generalização. Ele auxilia na escolha de parâmetros, ajustes no modelo e se necessário na troca da abordagem utilizada visando a melhor performance em previsões e observações futuras. O conjunto de teste é a porcentagem final usada após todo o processo de treinamento, para observar resultados a partir de dados nunca vistos fornecendo uma visão realista da sua capacidade de generalização.


**3. Explique como você lidaria com dados ausentes em um conjunto de dados de treinamento.**


Existem diferentes técnicas para abordar valores ausentes, são elas: Remoção de dados faltantes eliminando uma linha ou coluna, aplicavél quando a quantidade de dados ausentes é baixa; imputação pela média ou mediana; imputação da categoria mais frequente; imputação de valor constante defindo estrategicamente. Qualquer dessas técnicas seria aplicada após uma avaliação cuidadosa do dataset a ser trabalhado. Sua aplicação se daria somente no conjunto de treinamento.

**4. O que é uma matriz de confusão e como ela é usada para avaliar o desempenho de um modelo preditivo?**

A matriz de confusão fornece informações cruciais para a avaliação de um modelo, usada principalmente em classificação, especialemente binária. Ela é representada em forma de tabela trazendo as seguintes informações após o treinamento de um modelo: Verdadeiro Positivo (TP), Verdadeiro Negativo (TN), Falso Positivo (FP), Falso Negativo (FN). A partir da matriz de confusão podemos obter métricas fundamentais do desempenho de um modelo preditivo como:

- Accuracy (Acurácia)
  
  Mede o número total de classificações corretas dividido pelo número total de casos.

- Recall (Sensibilidade)

  Mede o número total de positivos verdadeiros dividido pelo número total de positivos reais.

- Precision (Precisão)

  Mede o número total de positivos verdadeiros dividido pelo número total de positivos previstos


- Specificity (Especificidade)

  Mede o número total de negativos verdadeiros dividido pelo número total de negativos reais.


- F1-Score

  É uma média harmônica de precision e recall
