1. Escreva uma função que receba uma lista de números e retorne outra lista
com os números ímpares.


```python
#função
def filtrarImpares(list):
  impares = []

  for num in list:
    if num % 2 != 0:
      impares.append(num)

  return impares
```

```python
#exemplo de uso

list = [1, 2, 3, 4, 5, 6, 7, 8, 9]

print(filtrarImpares(list))
```

2. Escreva uma função que receba uma lista de números e retorne outra lista
com os números primos presentes.

```python
#função para verificar se o número é primo
def eh_primo(num):
  if num <= 1:
    return False

  for div in range(2, num):
      if num % div == 0:
        return False

  return True


#função para adicionar número primo a lista
def filtrarPrimos(list):
  primos = []

  for num in list:
    if eh_primo(num):
      primos.append(num)

  return primos
```

```python
#exemplo de uso

list = [1, 2, 3, 4, 5, 6, 7, 8, 9]

primos = filtrarPrimos(list)

print(primos)
```

3. Escreva uma função que receba duas listas e retorne outra lista com os elementos que estão presentes em apenas uma das listas.


```python
#função para verificar números únicos nas listas

def elementosUnicos(list1, list2):
  elementos = []

  for num in list1:
    if num not in list2:
      elementos.append(num)

  for num in list2:
    if num not in list1:
      elementos.append(num)

  return elementos
```

```python
#Exemplo de uso

listA = [1, 2, 3, 5, 6, 8, 9]
listB = [0, 2, 3, 5, 6, 7, 8, 9]

unicos = elementosUnicos(listA, listB)
print(unicos)
```

4. Dada uma lista de números inteiros, escreva uma função para encontrar o segundo maior valor na lista.


```python
def segundoMaiorValor(list):
  maior = list[0]
  segundoMaior = list[0]

  for num in list:
    if num > maior:
      maior = num

    if num < maior and num > segundoMaior:
      segundoMaior = num

  return segundoMaior
```

```python
#exemplo de uso

list = [20, 60, 4, 23, 1]

print(f'o SEGUNDO MAIOR VALOR DA LISTA É {segundoMaiorValor(list)}')
```

5. Crie uma função que receba uma lista de tuplas, cada uma contendo o
nome e a idade de uma pessoa, e retorne a lista ordenada pelo nome das
pessoas em ordem alfabética.

```python
def listaOrdenadaNome(list):
  inOrder = sorted(list, key = lambda nome: nome[0])
  return inOrder
```

```python
#exemplo de uso

list = [
    ("Diego", 23),
    ("Bruno", 15),
    ("Aline", 65)
]

print(listaOrdenadaNome(list))
```

6. Como identificar e tratar outliers em uma coluna numérica usando desvio
padrão ou quartis?

```python
#USANDO DESVIO PADRÃO PARA IDENTIFICAR OUTLIER

#1 - Calcular média
#2 - Calcular desvio padrão
#3 - Definir limite inferior e superior

#EXEMPLO

import pandas as pd

df = pd.DataFrame({
    'idade': [22, 25, 27, 29, 30, 31, 32, 33, 35, 36, 38, 40, 42, 120]
})

print(df)

#Calculo media e desvio
media = df['idade'].mean()
desvio = df['idade'].std()

#fórmula para definir limites
limite_inferior = media - 3 * desvio
limite_superior = media + 3 * desvio

outliers_dp = df[(df['idade'] < limite_inferior) | (df['idade'] > limite_superior)]
print("OUTLIER IDENTIFICADO:")
print(outliers_dp)

```

```python
#USANDO DESVIO PADRÃO PARA IDENTIFICAR OUTLIER

#Quaris dividem os dados em 4 partes
#Q1 - 25%
#Q3 - 75%
#IQR = Q3 - Q1

#EXEMPLO

Q1 = df['idade'].quantile(0.25)
Q3 = df['idade'].quantile(0.75)
IQR = Q3 - Q1

lim_inf = Q1 - 1.5 * IQR
lim_sup = Q3 + 1.5 * IQR

outliers_iqr = df[(df['idade'] < lim_inf) | (df['idade'] > lim_sup)]
print(outliers_iqr)
```

É possível tratar Outliers Removendo-os, substituino pela média ou mediana ou mante-los caso sejam valores válidos

7. Como concatenar vários DataFrames (empilhando linhas ou colunas),
mesmo que tenham colunas diferentes?
Dica: Utiliza-se pd.concat() especificando axis=0 (linhas) ou axis=1
(colunas). Quando há colunas diferentes, os valores ausentes são
preenchidos com NaN.

```python
df1 = pd.DataFrame([1, 2, 3, 4, 5],
                  [10, 20, 30, 40, 50])

df2 = pd.DataFrame([6, 7, 8, 9, 10],
                   [60, 70, 80, 90, 100])

#Empilhar linhas (axis = 0)
pd.concat([df1, df2], axis=0)

```

```python
#Empilhar colunas (axis = 1)
pd.concat([df1, df2], axis=1)
```
